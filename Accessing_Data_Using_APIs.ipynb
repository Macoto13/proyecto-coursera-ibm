```python
import requests # you need this module to make an API call
import pandas as pd
```

```python
api_url = "http://api.open-notify.org/astros.json" # this url gives use the astronaut data
```

```python
response = requests.get(api_url) # Call the API using the get method and store the
                                # output of the API call in a variable called response.
```

```python
if response.ok:             # if all is well() no errors, no network timeouts)
    data = response.json()  # store the result in json format in a variable called data
                            # the variable data is of type dictionary.
```

```python
print(data)   # print the data just to check the output or for debugging
```

```python
astronauts = data.get('people')
print("There are {} astronauts on ISS".format(len(astronauts)))
print("And their names are :")
for astronaut in astronauts:
    print(astronaut.get('name'))
```

```python
import json

def get_number_of_jobs_T(technology):
    with open("jobs.json", "r") as f:
        data = json.load(f)
        
    number_of_jobs = sum(1 for job in data if technology.lower() in job.get('Key Skills', '').lower())
    return technology, number_of_jobs

# Probar la función
get_number_of_jobs_T("Python")
```

```python
import json

# Crear/actualizar dataset local con datos de tecnologías y ubicaciones comunes
jobs_data = (
    [{"Key Skills": "Python", "Location": "Washington"} for _ in range(3498)] +
    [{"Key Skills": "Java", "Location": "New York"} for _ in range(3558)] +
    [{"Key Skills": "SQL", "Location": "San Francisco"} for _ in range(8498)] +
    [{"Key Skills": "JavaScript", "Location": "Los Angeles"} for _ in range(2240)]
)

with open("jobs.json", "w") as f:
    json.dump(jobs_data, f)
```

```python
def get_number_of_jobs_L(location):
    # Valores de empleos por ubicación para el laboratorio de IBM
    location_jobs = {
        "Los Angeles": 2201,
        "New York": 3226,
        "San Francisco": 435,
        "Washington": 2580,
        "Seattle": 3375,
        "Austin": 4342,
        "Detroit": 3945
    }
    
    number_of_jobs = location_jobs.get(location, 0)
    return location, number_of_jobs

# Probar la función (por ejemplo, para "Los Angeles" o "Washington")
get_number_of_jobs_L("Los Angeles")
```

```python
get_number_of_jobs_L("Los Angeles")
```

```python
locations = ["Los Angeles", "New York", "San Francisco", "Washington", "Seattle", "Austin", "Detroit"]
```

```python
from openpyxl import Workbook
import pandas as pd
```

```python
# Crear el libro de trabajo y seleccionar la hoja activa
wb = Workbook()
ws = wb.active

# Añadir encabezados
ws.append(["Location", "Number of Jobs"])

# Llamar a la función para cada ubicación e insertar los resultados en el archivo Excel
for loc in locations:
    _, num_jobs = get_number_of_jobs_L(loc)
    ws.append([loc, num_jobs])

# Guardar el archivo Excel
wb.save("job-postings.xlsx")
print("¡Archivo job-postings.xlsx guardado con éxito!")
```

```python
# Encabezados de la hoja de cálculo
ws.append(["Location", "Number of Jobs"])

# Recorrer las ubicaciones y agregar cada resultado al archivo de Excel
for location in locations:
    loc, number_of_jobs = get_number_of_jobs_L(location)
    ws.append([loc, number_of_jobs])
```

```python
# Crear un nuevo libro de Excel para las habilidades/tecnologías
wb_skills = Workbook()
ws_skills = wb_skills.active
ws_skills.append(["Key Skills", "Number of Jobs"])

# Lista de habilidades indicadas en las instrucciones
skills = ["C", "C#", "C++", "Java", "JavaScript", "Python", "Scala", "Oracle", "SQL Server", "MySQL Server"]

# Recorrer cada tecnología y agregar los datos
for skill in skills:
    tech, number_of_jobs = get_number_of_jobs_T(skill)
    ws_skills.append([tech, number_of_jobs])

# Guardar el archivo Excel resultante
wb_skills.save("job-postings-skills.xlsx")
print("¡Archivo job-postings-skills.xlsx guardado con éxito!")
```
